Signal audit
A first-pass exploratory check of each candidate feature: distribution, missingness, and linear correlation with the decline label, each with a plain-language verdict. This complements (rather than duplicates) the Random Forest feature importance results in w05_model.ipynb, since correlation captures linear relationships while the trained model captures non-linear ones.

```python
import os
import pandas as pd
import numpy as np

if not os.path.exists("/content/flyrank-ml"):
    !git clone https://github.com/haroonrana330/flyrank-ml.git /content/flyrank-ml
os.chdir("/content/flyrank-ml")

url = "https://raw.githubusercontent.com/haroonrana330/flyrank-ml/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

df["is_declining_label"] = (df["trend_direction"].astype(str).str.lower() == "down").astype(int)

signal_cols = ["search_volume", "competition", "cpc", "word_count", "char_count",
               "ctr", "avg_position", "engagement_rate", "scroll_rate", "ai_traffic_pct"]

for col in signal_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

print("=" * 70)
print("SIGNAL AUDIT: univariate stats per feature")
print("=" * 70)
print(df[signal_cols].describe().T[["mean", "std", "min", "max"]])

print()
print("=" * 70)
print("SIGNAL AUDIT: missingness per feature")
print("=" * 70)
missing = df[signal_cols].isna().mean().sort_values(ascending=False)
print((missing * 100).round(2).astype(str) + "%")

print()
print("=" * 70)
print("SIGNAL AUDIT: correlation with the decline label")
print("=" * 70)
corrs = df[signal_cols + ["is_declining_label"]].corr()["is_declining_label"].drop("is_declining_label")
corrs_sorted = corrs.sort_values(key=abs, ascending=False)
print(corrs_sorted.round(4))

print()
print("=" * 70)
print("VERDICTS")
print("=" * 70)
for col, val in corrs_sorted.items():
    strength = "weak"
    if abs(val) >= 0.15:
        strength = "moderate"
    if abs(val) >= 0.3:
        strength = "strong"
    direction = "positively" if val > 0 else "negatively"
    print(f"{col}: {strength} signal, correlates {direction} with decline (r={val:.3f})")

print()
print("Note: these are simple linear correlations for a first-pass signal check.")
print("The Week-5 Random Forest model (w05_model.ipynb) captures non-linear")
print("relationships between these same signals, which is why its feature")
print("importance ranking (avg_position dominant) differs somewhat from the")
print("linear correlation ranking here -- both are valid, complementary views.")
```
